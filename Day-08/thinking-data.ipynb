```python
import json
```

```python
# load the data (json file)

def loaddata(file):
    with open(file, "r") as f:
        data= json.load(f) 

    return data
```

```python
data =loaddata("store_data.json")
```

```python
#clean and structure the data

def clean_data(data):
    cleaned_data = []
    seen_names = set()

    for item in data:
        cleaned_item = item.copy()

        rating_map = {
            "one": 1,
            "two": 2,
            "three": 3,
            "four": 4,
            "five": 5
        }

        rating = cleaned_item.get("rating")

        if isinstance(rating, str):
            rating = rating.lower().strip()

            if rating in rating_map:
                cleaned_item["rating"] = rating_map[rating]

        if "feedback" in cleaned_item:
            cleaned_item["feedback"] = (
                cleaned_item["feedback"].strip().capitalize()
            )

        if "age" not in cleaned_item:
            cleaned_item["age"] = None

        # Remove duplicate names
        if cleaned_item["name"] not in seen_names:
            cleaned_data.append(cleaned_item)
            seen_names.add(cleaned_item["name"])

    return cleaned_data
```

```python
cleaned_data =clean_data(data)
```

```python
# Average Rating

ratings = []

for item in cleaned_data:
    ratings.append(item["rating"])

avg_rating = sum(ratings) / len(ratings)


# Poor Rating (<3)

poor_rating = 0

for item in cleaned_data:
    if item["rating"] < 3:
        poor_rating += 1

poor_percentage = (poor_rating / len(cleaned_data)) * 100


print("Average Rating:", avg_rating)
print("Poor Rating Percentage:", poor_percentage, "%")
```
